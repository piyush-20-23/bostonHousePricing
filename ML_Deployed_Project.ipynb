Feature Selection

```python
## Independent and Dependent features

X = df.iloc[:, :-1]
y = df.iloc[: , -1]
```

```python
X.columns
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state = 42)
```

```python
X_train
```

```python
X_test
```

Standardize the dataset

```python
from sklearn.preprocessing import StandardScaler
scalar = StandardScaler()
```

```python
X_train = scalar.fit_transform(X_train)
X_test = scalar.transform(X_test)
# do not use fit_transform as to ensure that the same technique is applied to the test
```

```python
import pickle 
pickle.dump(scalar, open('scaling.pkl', 'wb'))
```

# Model Training

```python
from sklearn.linear_model import LinearRegression
```

```python
regression = LinearRegression()
regression.fit(X_train, y_train)
```

```python
print(regression.coef_)
```

```python
print(regression.intercept_)
```

```python
## on which parameter the model has been trained
regression.get_params()
```

```python
## prediction with test data
reg_pred = regression.predict(X_test)
print(reg_pred)
```

```python
## plot a scatter plot for prediction
plt.scatter(y_test, reg_pred)
plt.xlabel("Orignal data")
plt.ylabel("Predicted data")
```

```python
## prediction with residuals(error)
residuals = y_test - reg_pred
```

```python
residuals
```

```python
average = sum(residuals) / len(residuals)
print(average)
```

```python
## ploting residuals
sns.displot(residuals, kind = "kde")
```

```python
## scatter plot with prediction and residuals
plt.scatter(reg_pred, residuals)
# this graph gives uniform distribution implies there is no relation between residuals and the value on which they were created
```

# Metric evaluation

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error

print(mean_absolute_error(y_test, reg_pred))
```

```python
print(mean_squared_error(y_test, reg_pred))
```

```python
print(np.sqrt(mean_squared_error(y_test, reg_pred)))
```

```python
# r square and adjusted r square

#                SSR (sum of squares of residuals)
#    R^2 = 1 - --------------------------------------
#                SST (total sum of squares)

#
#    Adjusted R^2 = 1 - [(1 - R2)*(n-1) / (n-k-1)]
#
# n : number of observation
# k : the number of predictor variable
```

```python
from sklearn.metrics import r2_score
score = r2_score(y_test, reg_pred)
print(score)
# r2_score near to 1 is good
```

```python
adjusted_R2 = 1 - ((1 - score)*(len(y_test)) / (len(y_test) - X_test.shape[1] - 1))
print(adjusted_R2)
```
